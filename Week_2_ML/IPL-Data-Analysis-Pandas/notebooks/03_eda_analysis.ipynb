# 03 EDA Analysis

Simple tables for team wins, matches per season, and top venues.

```python
import pandas as pd

df = pd.read_csv('../data/matches.csv')
df.columns = [c.strip() for c in df.columns]
df = df.dropna(subset=['winner']).copy()
df = df[['season', 'winner', 'venue', 'toss_decision']]

team_wins = df['winner'].value_counts()
matches_per_season = df.groupby('season').size().sort_index()
top_venues = df['venue'].value_counts()

team_wins.head(10)
```
