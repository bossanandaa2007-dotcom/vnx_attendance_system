# 02 Data Cleaning

Clean only the columns required for analysis.

```python
import pandas as pd

df = pd.read_csv('../data/matches.csv')
df.columns = [c.strip() for c in df.columns]

before_rows = len(df)
df_clean = df.dropna(subset=['winner']).copy()
after_rows = len(df_clean)

keep_cols = ['season', 'winner', 'venue', 'toss_decision']
df_clean = df_clean[keep_cols]

print('Total rows before cleaning:', before_rows)
print('Total rows after cleaning:', after_rows)
print('Missing values:', df_clean.isna().sum().sum())

df_clean.head()
```
