```python
import pandas as pd

# Load cleaned data
df = pd.read_csv("data/cleaned_data.csv")

# 1. Average marks
avg_marks = df["marks"].mean()
print("Average Marks:", avg_marks)

# 2. Students scoring above average
above_avg_students = df[df["marks"] > avg_marks]
print("\nStudents scoring above average:")
print(above_avg_students)

# 3. City-wise average marks
city_avg = df.groupby("city")["marks"].mean()
print("\nCity-wise average marks:")
print(city_avg)

```
